```python
#Variablen: 

string = "Ich bin ein String"
liste = ["Ich", "bin", "eine", "Liste"]
zahl = 5
```

```python
print(string)
print(liste)
print(zahl)
```

```python
#if-else-Abfrage: 

if zahl == 3:
    print("Die Zahl ist 3!")
elif zahl <3: 
    print("Die Zahl ist kleiner als 3")
else:
    print("Die Zahl ist größer als 3")
```

```python
#for-Schleifen: 

for item in liste:
    print(item)
```
